```python
productos = []
while True:
    nombre = input("Ingrese producto (o 'fin' para salir): ")
    if nombre == "fin":
        break
    precio = float(input("Precio: "))
    # Usa continue para saltar productos con precio negativo o cantidad menor a 0.
    if precio < 0:
        print("Precio inválido. Debe ser positivo.")
        continue
    cantidad = int(input("Cantidad: "))
    if cantidad < 0:
        print("Cantidad inválida. Debe ser mayor o igual que cero.")
        continue
    productos.append({"nombre": nombre, "precio": precio, "cantidad": cantidad})

if not productos:
  print("No se ingresaron productos")

else:
  # Mostrar total
  total = 0
  for p in productos:
      total += p["precio"] * p["cantidad"]

  print("Total del inventario:", total)

  #Muestra el producto más caro usando max().
  def obtener_precio (producto):
    return producto["precio"]
  producto_caro = max(productos, key = obtener_precio)
  print(f"Producto más caro: {producto_caro}")

  # Representa el estado final del inventario como lista de diccionarios.
  print(productos)

```
