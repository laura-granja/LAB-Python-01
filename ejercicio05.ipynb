```python
### Juego de adivinanza
import random

def jugar_partida():
  numero = random.randint(1, 100)
  intentos = []
  gano = False
  while len(intentos) < 7 and not gano:
    numero_ingresado = input("\nIngrese un numero: ")

    try:
      numero_ingresado = int(numero_ingresado)
    except ValueError:
      print("Ingrese un numero valido")
      continue

    if numero_ingresado > 100 or numero_ingresado < 1:
      print("Ingrese un numero valido")
      continue
    if numero_ingresado in intentos:
      print("Numero ingresado anteriormente")
      continue
    intentos.append(numero_ingresado)

    if numero_ingresado == numero:
      gano = True
    else:
      if numero_ingresado < numero:
        print("\nEl número es MAYOR al número ingresado")
      else:
        print("\nEl número es MENOR al número ingresado")
      print(f"\nIntentos restantes: {7 - len(intentos)}")

  if gano:
    print("\nGanaste la partida")
  else:
    print(f"\nPerdiste la partida, el número era: {numero}")

  print("Historial de Intentos:", end= " ")
  for intento in intentos:
    print(intento, end= " ")
  print()
  return gano

ganadas = 0
perdidas = 0
jugar_nuevamente = True

while jugar_nuevamente:
  if jugar_partida():
    ganadas += 1
  else:
    perdidas += 1

  print(f"\nPartidas ganadas: {ganadas}")
  print(f"Partidas perdidas: {perdidas}")

  rta = input("¿Quiere jugar nuevamente? (s/n) ").strip().lower()
  while rta != "s" and rta != "n":
    rta = input("Responda 's' o 'n': ").strip().lower()
  jugar_nuevamente = (rta == "s")

print("Gracias por jugar")
```

#¿Por qué 7 intentos son suficientes si se juega de forma óptima?

Siete intentos son suficientes si se juega de forma óptima, porque la estrategia óptima es una búsqueda binaria. En cada intento se adivina el número que está en la mitad del rango posible, y la respuesta (“mayor” o “menor”) permite descartar la otra mitad.
En el peor caso, la cantidad de intentos necesarios es ⌈log₂ 100⌉ = 7, porque se utiliza busqueda binaria. Es decir que con los 7 intentos alcanza a cubrir todos los números entre los 100.


```python
import random
TAM = 10

def crear_tablero_aleatorio():
  tablero = []
  for i in range(TAM):
    fila = []
    for j in range(TAM):
      fila.append(random.randint(0, 1))
    tablero.append(fila)
  return tablero

def crear_tablero_vacio():
  tablero = []
  for i in range(TAM):
    fila = []
    for j in range(TAM):
      fila.append(0)
    tablero.append(fila)
  return tablero

def celda_viva(tablero, fila, columna):
  vivos = 0
  if fila < TAM and columna < TAM and fila >= 0 and columna >= 0:
        if tablero[fila][columna] == 1:
          vivos +=1
  return vivos

def contar_vecinos_vivos(tablero, fila, columna):
  vivos = 0
  for mov_fil in (-1, 0, 1):
    for mov_col in (-1, 0, 1):
      if mov_fil == 0 and mov_col == 0:
        continue
      fil = fila + mov_fil
      col = columna + mov_col
      if 0 <= fil and fil < TAM and 0 <= col and col < TAM:
        vivos += tablero[fil][col]
  return vivos

def siguiente_generacion(tablero):
  tablero_nuevo = crear_tablero_vacio()
  for i in range(TAM):
    for j in range(TAM):
      vecinos = contar_vecinos_vivos(tablero, i, j)
      if tablero[i][j] == 1:
        if vecinos < 2 or vecinos > 3:
          tablero_nuevo[i][j] = 0
        else:
          tablero_nuevo[i][j] = 1
      else:
        if vecinos == 3:
          tablero_nuevo[i][j] = 1
        else:
          tablero_nuevo[i][j] = 0
  return tablero_nuevo

def imprimir_tablero(tablero):
  for fila in tablero:
    for celda in fila:
      if celda == 1:
        print("■", end= " ")
      else:
        print("·", end= " ")
    print()

tablero = crear_tablero_aleatorio()
for generacion in range (5):
  print(f"Generación {generacion + 1}")
  imprimir_tablero(tablero)
  tablero = siguiente_generacion(tablero)
  print()
```

#Tablero nuevo

Es necesario usar un tablero nuevo porque las reglas del juego de la vida se aplican a todas las celdas al mismo tiempo, y el estado de cada celda en la siguiente generación depende solo de los vecinos que tenía en la generación actual.

Si se modificara el mismo tablero mientras se recorre, las celdas ya actualizadas se mezclarían con las de la generación anterior, y al contar los vecinos de una celda anterior, algunos ya tendrían su valor nuevo y al contar los vecinos daría un resultado incorrecto.
