```python
productos = []

while True:
    nombre = input("Ingrese producto (o 'fin' para salir): ")
    if nombre == "fin":
        break
    # Proteger la lectura
    while True:
      try:
        precio = float(input("Precio: "))
      except ValueError:
        print("Precio inválido")
        continue
      else:
        if precio <= 0:
          print("Precio inválido")
          continue
        break
    # 'else' solo se ejecuta si el 'try' terminó sin errores
    while True:
      try:
        cantidad = int(input("Cantidad: "))
      except ValueError:
        print("Cantidad inválida")
        continue
      else:
        if cantidad < 0:
          print("Cantidad inválida")
          continue
        break

    """
    No sé utilizo 'finally',
    porque su función es ejecutar instrucciones que deben realizarse independientemente de si ocurrió un error.
    Por lo que en este caso no es necesario
    """
    productos.append({"nombre": nombre, "precio": precio, "cantidad": cantidad})

# Mostrar total
total = 0
for p in productos:
    total += p["precio"] * p["cantidad"]

print("Total del inventario:", total)
```

```python

import csv
import json

def guardar_csv(productos):
  archivo = open("inventario.csv", "w")
  archivo.write("nombre,precio,cantidad\n")
  for producto in productos:
    linea = producto["nombre"] + ", " + str(producto["precio"]) + ", " + str(producto["cantidad"]) + "\n"
    archivo.write(linea)
  archivo.close()
  print("Inventario guardado")

def cargar_csv():
  productos = []
  try:
    archivo = open("inventario.csv", "r")
    archivo.readline()
    for linea in archivo:
      datos = linea.strip().split(", ")
      productos.append({"nombre": datos[0], "precio": float(datos[1]), "cantidad": int(datos[2])})
    archivo.close()
  except FileNotFoundError:
    print("El archivo no existe. Inventario vacío")
  return productos

def guardar_json(productos):
  with open("inventario.json", "w") as archivo:
    json.dump(productos, archivo, indent=2)
  print("Inventario guardado")

def cargar_json():
  try:
    with open("inventario.json", "r") as archivo:
      productos = json.load(archivo)
  except FileNotFoundError:
    print("El archivo no existe. Inventario vacío")
    productos = []
  return productos

guardar_csv(productos)
inventario_csv = cargar_csv()

print("Inventario recuperado desde CSV:")
print(inventario_csv)

guardar_json(productos)
inventario_json = cargar_json()

print("Inventario recuperado desde JSON:")
print(inventario_json)
```

# ¿Qué ventajas tiene frente al CSV?
Json tiene la ventaja de que permite guardar estructuras de datos como listas y diccionarios, conservando su orden y los tipos de datos. Además, mediante json.dump() y json.load() se puede guardar y recuperar la información sin tener que separar y convertir cada dato, como se hace con csv, json lo hace automáticamente
