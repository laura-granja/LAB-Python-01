```python
# Estado inicial
saldo = float(input("Ingrese el saldo de la cuenta"))
deposito = float(input("Ingrese monto a depositar: "))
retiro = float(input("Ingrese monto a retirar: "))

# Actualización del estado
saldo = saldo + deposito   # Se deposita dinero
saldo = saldo - retiro     # Se retira dinero

# Salida del estado final
print("Saldo final de la cuenta:", saldo)

interes = 0.02
saldo_final = saldo + saldo * interes
print(f"Saldo final con interés: ${saldo_final:.2f}")
```

### Traza del estado

Valores de ejemplo ingresados: saldo = 500000 , depósito = 150000 , retiro = 80000

| Instrucción | saldo | deposito | retiro | interes | saldo_final |
|---          |---    |---       |---     |---      |---          |
| `saldo = float(input(...))` (crea `saldo`)|500000.0  | no existe | no existe | no existe | no existe |
| `deposito = float(input(...))` (crea `deposito`)|500000.0  |150000.0  | no existe | no existe | no existe |
| `retiro = float(input(...))` (crea `retiro`)|500000.0  |150000.0  |80000.0  | no existe | no existe |
| `saldo = saldo + deposito` |650000.0  |150000.0  |80000.0  | no existe | no existe |
| `saldo = saldo - retiro` |570000.0  |150000.0  |80000.0  | no existe | no existe |
| `interes = 0.02` (crea `interes`)|570000.0  |150000.0  |80000.0  |0.02  | no existe |
| `saldo_final = saldo + saldo * interes` (crea `saldo_final`)|570000.0  |150000.0  |80000.0  |0.02  |581400.00  |

### Estado inicial y final

- **Estado inicial:** antes de ejecutar la primera instrucción la memoria del programa está vacía, es decir, no existe ninguna variable. Los valores aparecen a medida que se ejecutan las asignaciones.
- **Estado final:** al terminar el programa, la memoria contiene las siguientes variables: `saldo = 570000.0`, `deposito = 150000.0`, `retiro = 80000.0`, `interes = 0.02` y `saldo_final = 581400.00`. Se puede ver que `saldo` conserva el valor posterior al depósito y al retiro, mientras que `saldo_final` es una variable distinta que guarda el resultado de aplicar el interés.

### Cómo afecta cada instrucción a la memoria

- `saldo = float(input(...))`, `deposito = float(input(...))` y `retiro = float(input(...))`: `input()` lee lo que escribe el usuario como texto (`str`), `float()` lo convierte a número decimal y la asignación crea la variable en memoria y guarda ese valor.
- `saldo = saldo + deposito`: primero se evalúa la expresión de la derecha con los valores actuales (500000.0 + 150000.0) y luego el resultado (650000.0) sobrescribe el valor anterior de `saldo`, que se pierde.
- `saldo = saldo - retiro`: funciona igual: se evalúa 650000.0 - 80000.0 y el resultado (570000.0) reemplaza nuevamente el valor de `saldo`.
- `interes = 0.02`: crea una nueva variable de tipo `float` con el valor del interés mensual.
- `saldo_final = saldo + saldo * interes`: lee los valores de `saldo` e `interes` sin modificarlos y crea una nueva variable `saldo_final` con el resultado (581400.00).
