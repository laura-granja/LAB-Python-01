```python
# Estado inicial
saldo = float(input("Ingrese el saldo de la cuenta"))
deposito = float(input("Ingrese monto a depositar: "))
retiro = float(input("Ingrese monto a retirar: "))

# Actualización del estado
saldo = saldo + deposito   # Se deposita dinero
saldo = saldo - retiro     # Se retira dinero

# Salida del estado final
print("Saldo final de la cuenta:", saldo)

interes = 0.02
saldo_final = saldo + saldo * interes
print(f"Saldo final con interés: ${saldo_final:.2f}")
```

### Traza del estado

Valores de ejemplo ingresados: saldo = 500000 , depósito = 150000 , retiro = 80000

| Instrucción | saldo | deposito | retiro | interes | saldo_final |
|---          |---    |---       |---     |---      |---          |
| `saldo = float(input(...))` (crea `saldo`)|500000.0  | no existe | no existe | no existe | no existe |
| `deposito = float(input(...))` (crea `deposito`)|500000.0  |150000.0  | no existe | no existe | no existe |
| `retiro = float(input(...))` (crea `retiro`)|500000.0  |150000.0  |80000.0  | no existe | no existe |
| `saldo = saldo + deposito` |650000.0  |150000.0  |80000.0  | no existe | no existe |
| `saldo = saldo - retiro` |570000.0  |150000.0  |80000.0  | no existe | no existe |
| `interes = 0.02` (crea `interes`)|570000.0  |150000.0  |80000.0  |0.02  | no existe |
| `saldo_final = saldo + saldo * interes` (crea `saldo_final`)|570000.0  |150000.0  |80000.0  |0.02  |581400.00  |

### Estado inicial y final

- **Estado inicial:** antes de ejecutar la primera instrucción la memoria del programa está vacía, es decir, no existe ninguna variable. Los valores aparecen a medida que se ejecutan las asignaciones.
- **Estado final:** al terminar el programa, la memoria contiene las siguientes variables: `saldo = 570000.0`, `deposito = 150000.0`, `retiro = 80000.0`, `interes = 0.02` y `saldo_final = 581400.00`. Se puede ver que `saldo` conserva el valor posterior al depósito y al retiro, mientras que `saldo_final` es una variable distinta que guarda el resultado de aplicar el interés.

### Cómo afecta cada instrucción a la memoria

- `saldo = float(input(...))`, `deposito = float(input(...))` y `retiro = float(input(...))`: `input()` lee lo que escribe el usuario como texto (`str`), `float()` lo convierte a número decimal y la asignación crea la variable en memoria y guarda ese valor.
- `saldo = saldo + deposito`: primero se evalúa la expresión de la derecha con los valores actuales (500000.0 + 150000.0) y luego el resultado (650000.0) sobrescribe el valor anterior de `saldo`, que se pierde.
- `saldo = saldo - retiro`: funciona igual: se evalúa 650000.0 - 80000.0 y el resultado (570000.0) reemplaza nuevamente el valor de `saldo`.
- `interes = 0.02`: crea una nueva variable de tipo `float` con el valor del interés mensual.
- `saldo_final = saldo + saldo * interes`: lee los valores de `saldo` e `interes` sin modificarlos y crea una nueva variable `saldo_final` con el resultado (581400.00).

```python
# Lista de productos en una tienda
inventario = [
    {"nombre": "Pan", "precio": 1500, "stock": 30},
    {"nombre": "Leche", "precio": 3500, "stock": 15},
    {"nombre": "Café", "precio": 8000, "stock": 10},
]

# Muestra el nombre del primer producto.
print(f"Nombre del primer producto: {inventario[0]['nombre']}")

# Agrega un nuevo producto al inventario.
inventario.append({"nombre": "Huevos", "precio": 700, "stock": 0})

# Disminuye el stock de “Leche” en 2 unidades (venta).
inventario[1]["stock"] -= 2

# Calcula el valor total del inventario (precio * stock por producto).
valor_total = 0
for producto in inventario:
  valor_total += producto["precio"] * producto["stock"]

# Muestra los productos cuyo stock sea menor a 10.
for producto in inventario:
  if producto["stock"] < 10:
    print(producto['nombre'])

# Aumenta el precio de todos los productos en un 10%.
aumento = 0.1
for producto in inventario:
  producto["precio"] *= (1 + aumento)

# Redondea los precios con round(p["precio"], 2) para evitar decimales largos.
for producto in inventario:
  producto["precio"] = round(producto["precio"], 2)

# Busca un producto por nombre (ingresado por el usuario) y muestra el producto encontrado, ten en cuenta que no puede estar.
buscar_nombre = input("Ingrese el nombre del producto a buscar: ")
buscar_nombre = buscar_nombre.lower()
producto_encontrado = None
for producto in inventario:
  if producto['nombre'].lower() == buscar_nombre:
    producto_encontrado = producto
    break
if producto_encontrado:
  print(f"Producto encontrado: {producto_encontrado}")
else:
  print("Producto no encontrado")

# Elimina un producto del inventario si su stock es 0.
productos_a_eliminar = []
for producto in inventario:
  if producto["stock"] == 0:
    productos_a_eliminar.append(producto)
for producto in productos_a_eliminar:
  inventario.remove(producto)

# Ordena la lista por precio ascendente.
def ordenar_por_precio(producto):
  return producto["precio"]
inventario.sort(key=ordenar_por_precio)

# Muestra el inventario actualizado.
print(f"Inventario actualizado: {inventario}")
```

```python
# Tupla: coordenadas (x, y)
punto = (3, 7)
print("Punto:", punto)

# Crea una tupla rectangulo = (ancho, alto).
rectangulo = (5, 10)

# Calcula su área y perímetro.
ancho, alto = rectangulo
area = ancho * alto
perimetro = 2 * (ancho + alto)

print(f"Área del rectángulo: {area}")
print(f"Perímetro del rectángulo: {perimetro}")

"""
Intenta cambiar el valor de ancho.
rectangulo[0] = 4
¿Qué error obtienes?
TypeError: 'tuple' object does not support item assignment
¿Por qué?
Porque una tupla es inmutable, es decir, que una vez creada no puede cambiar su contenido.
"""
posiciones = [(0, 0), (1, 2), (3, 4)]

# Calcula la distancia total recorrida dadas las posiciones anteriores
distancia_total = 0
for i in range(len(posiciones)):
  x, y = posiciones[i]
  if i > 0:
    x_prev, y_prev = posiciones[i - 1]
    distancia = ((x - x_prev) ** 2 + (y - y_prev) ** 2) ** 0.5
  distancia_total += distancia
print(f"Distancia total recorrida: {distancia_total}")

```

```python
biblioteca = {
    "libros": [
        {"titulo": "1984", "autor": "George Orwell", "prestado": False},
        {"titulo": "El Principito", "autor": "Antoine de Saint-Exupéry", "prestado": True},
    ],
    "usuarios": ["Ana", "Luis", "María"]
}

# Muestra cuántos libros están prestados y cuántos disponibles.
if biblioteca["libros"]:
  libros_prestados = 0
  libros_disponibles = 0
  for libro in biblioteca["libros"]:
    if libro["prestado"] == True:
      libros_prestados += 1
    else:
      libros_disponibles += 1
  print(f"Libros prestados: {libros_prestados}")
  print(f"Libros disponibles: {libros_disponibles}")

# Agrega un nuevo libro
biblioteca["libros"].append({"titulo": "Cien años de soledad", "autor": "Gabriel Garcia", "prestado": False})

# Marca como prestado el libro “1984”.
biblioteca["libros"][0]["prestado"] = True

# Muestra el estado actualizado del sistema (usa print(biblioteca)).
print(biblioteca)
```
