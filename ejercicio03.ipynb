```python
import requests

# Llamado a una API pública
url = "https://jsonplaceholder.typicode.com/users"
response = requests.get(url)
usuarios = response.json()
# Procesamiento imperativo
nombres = []
for u in usuarios:
    nombres.append(u["name"])
print("Usuarios encontrados:")
for n in nombres:
    print("-", n)

# Filtrar solo los usuarios cuyo correo termine en .biz
usuarios_biz = []
for u in usuarios:
  if u["email"].endswith(".biz"):
    usuarios_biz.append(u)
# Mostrar el nombre de estos usuarios
print("Usuarios con correo .biz:")
for u in usuarios_biz:
    print("-", u["name"])
```

```python
import requests

# Entrada de datos
nombre_pokemon = input("Ingrese el nombre de un Pokémon: ").lower()

# Llamado a la API
url = f"https://pokeapi.co/api/v2/pokemon/{nombre_pokemon}"
respuesta = requests.get(url)


# Verificación del estado de la respuesta
if respuesta.status_code == 200:
    datos = respuesta.json()

    # Reconstrucción de tipos y habilidades sin comprensiones de listas
    tipos = []
    for t in datos["types"]:
      tipos.append(t["type"]["name"])

    habilidades = []
    for h in datos["abilities"]:
      habilidades.append(h["ability"]["name"])

    estadisticas = {}
    for e in datos["stats"]:
      estadisticas[e["stat"]["name"]] = e["base_stat"]

    # Modelado del estado del Pokémon, con la modificación de tipos y habilidades y un nuevo uso (Estadísticas)
    pokemon = {
        "nombre": datos["name"],
        "altura": datos["height"],
        "peso": datos["weight"],
        "tipos": tipos,
        "habilidades": habilidades,
        "estadisticas": estadisticas
        }

    # Si el Pokémon tiene más de un tipo, mostrar "Pokémon de tipo mixto".
    if len(tipos) > 1:
      print(f"El pokémon {pokemon['nombre'].capitalize()} es de tipo mixto")

    # Si el peso es mayor a 100, mostrar "Pokémon de gran tamaño"
    if pokemon["peso"] > 100:
      print(f"El pokémon {pokemon['nombre'].capitalize()} es de gran tamaño")
    # En caso contrario, "Pokémon ligero"
    else:
      print(f"El pokémon {pokemon['nombre'].capitalize()} es ligero")

    # Salida del estado actual
    print("\n📘 Información del Pokémon")
    print(f"Nombre: {pokemon['nombre'].capitalize()}")
    print(f"Altura: {pokemon['altura']}")
    print(f"Peso: {pokemon['peso']}")
    print(f"Tipos: {', '.join(pokemon['tipos'])}")
    # Modificación usando for para listar las habilidades una por una
    print("Habilidades:")
    for habilidad in habilidades:
      print(f"- {habilidad}")
    # Imprimir las estadísticas listadas por su nombre y con su valor
    print("Estadísticas del Pokémon")
    for estadistica, valor in pokemon["estadisticas"].items():
      print(f"- {estadistica.capitalize()}: {valor}")
    # Estadística que más se destaca el Pokemon
    for estadistica, valor in estadisticas.items():
      valor_max = max(estadisticas.values())
      if valor == valor_max:
        print(f"Se destaca en {estadistica.capitalize()}")

else:
    print("Pokémon no encontrado. Verifique el nombre ingresado.")
```
