**Error obtenido:** `UnboundLocalError: cannot access local variable 'contador_global' where it is not associated with a value`

**Por qué ocurre:** dentro de la función `incrementar()` la instrucción `contador_global = contador_global + 1` es una asignación, así que Python trata `contador_global` como una variable local de la función, distinta de la variable global del mismo nombre. Al evaluar la expresión de la derecha, esa variable local todavía no tiene ningún valor, y por eso se produce el error. Además, aunque no fallara, la función modificaría solo su variable local y el valor global seguiría siendo 0.

```python
contador_global = 0

def incrementar():
  global contador_global
  contador_global = contador_global + 1

incrementar()
print(contador_global)
```

```python
contador_global = 0

def incrementar(contador_global):
    contador_global = contador_global + 1
    return contador_global

contador_global = incrementar(contador_global)
print(contador_global)
```

**¿Cuál versión es más fácil de probar y por qué?**

La versión que recibe el valor como parámetro y lo retorna es más fácil de probar. Su resultado solo depende del argumento que recibe y no modifica nada fuera de la función, por lo que basta con llamarla con un valor conocido (por ejemplo, `incrementar(5)`) y comprobar que devuelve el esperado (`6`)

En cambio, la versión con `global` modifica una variable externa, por lo que su resultado depende del estado en que se encuentre esa variable en el momento de la llamada. Para probarla hay que dejar el contador en un valor conocido antes de cada prueba, y el resultado cambia según cuántas veces se haya llamado antes.

La versión con `global` es más corta y sencilla de escribir, pero tiene efectos secundarios y depende del estado, lo que dificulta probarla

# Modificar un objeto mutable vs. reasignar la variable local
* notas.append(nueva) modifica el objeto. El parámetro notas y mis_notas son dos nombres que apuntan a la misma lista en la memoria. Como append cambia el contenido de ese objeto, el cambio se ve afuera de la función.
* notas = [] reasigna el nombre. Esta instrucción crea una lista vacía nueva y hace que la variable local notas apunte a ella. La lista original no se toca y mis_notas sigue apuntando a la original, sin modificarla. Al terminar la función, la variable local desaparece junto con su lista vacía

```python
# Uso de id()
def agregar_nota(notas, nueva):
    antes = id(notas)
    print(f"dentro de agregar_nota, antes: {antes}")
    notas.append(nueva)
    despues = id(notas)
    print(f"dentro de agregar_nota, después: {despues}")
    if antes == despues:
        print("notas no ha cambiado, apunta al mismo objeto")
    else:
        print("notas ha cambiado, apunta a diferentes objetos")

def reiniciar(notas):
    antes = id(notas)
    print(f"dentro de reiniciar, al entrar: {antes}")
    notas = []
    despues = id(notas)
    print(f"dentro de reiniciar, tras notas = []: {despues}")
    if antes == despues:
        print("notas no ha cambiado, apunta al mismo objeto")
    else:
        print("notas ha cambiado, apunta a diferentes objetos")

def reiniciar_og(notas):
    antes = id(notas)
    print(f"dentro de reiniciar_og, al entrar: {antes}")
    notas.clear()
    despues = id(notas)
    print(f"dentro de reiniciar_og, tras clear(): {despues}")
    if antes == despues:
        print("notas no ha cambiado, apunta al mismo objeto")
    else:
        print("notas ha cambiado, apunta a diferentes objetos")

mis_notas = [3.5, 4.0]
print(f"id de mis_notas: {id(mis_notas)}")
agregar_nota(mis_notas, 4.8)
reiniciar(mis_notas)
despues_reiniciar = id(mis_notas)
print(f"id de mis_notas tras reiniciar: {despues_reiniciar}")
reiniciar_og(mis_notas)
despues_reiniciar_og = id(mis_notas)
print(f"id de mis_notas tras reiniciar_og: {despues_reiniciar_og}")
```

```python
cuenta = {"titular": "Ana", "saldo": 200_000, "movimientos": []}

def menu():
  print("\n¿Qué desea hacer?")
  print("1. Consultar saldo")
  print("2. Depositar dinero")
  print("3. Retirar dinero")
  print("Si desea finalizar la operación, escriba 'salir'")
  opcion = input("\nEscriba solamente el número de la opción que desea realizar: ").strip().lower()
  return opcion

def operacion(cuenta, opcion):
  while opcion != "salir":
    if opcion == "1":
      consultar(cuenta)
    elif opcion == "2":
      monto = leer_monto(input("Escriba el monto que desea ingresar a la cuenta: "))
      if monto is not None:
        depositar(cuenta, monto)
    elif opcion == "3":
      monto = leer_monto(input("Escriba el monto que desea retirar de la cuenta: "))
      if monto is not None:
        retirar(cuenta, monto)
    else:
      print("Opción inválida")
    opcion = menu()
  mostrar_extracto(cuenta)
  print("Gracias por usar el cajero")

def leer_monto(monto):
  try:
    monto = int(monto)
  except ValueError:
    print("El monto debe ser un número entero")
    return None
  if monto <= 0:
    print("El monto debe ser mayor que 0")
    return None
  return monto

def consultar(cuenta):
  print(f"El saldo de la cuenta es de ${cuenta['saldo']:,}")
  cuenta['movimientos'].append({"tipo": "Consulta", "monto": None , "saldo resultante": cuenta['saldo']})

def depositar(cuenta, monto):
  cuenta['saldo'] += monto
  cuenta['movimientos'].append({"tipo": "Deposito", "monto": monto, "saldo resultante": cuenta['saldo']})

def retirar(cuenta, monto):
  if monto <= cuenta['saldo']:
    cuenta['saldo'] -= monto
    cuenta['movimientos'].append({"tipo": "Retiro", "monto": monto, "saldo resultante": cuenta['saldo']})
  else:
    print("Fondos insuficientes")

def mostrar_extracto(cuenta):
  print(f"\nExtracto de movimientos - {cuenta['titular']}")
  if not cuenta["movimientos"]:
    print("No hubo movimientos")
    return

  linea = "+" + "-" * 12 + "+" + "-" * 16 + "+" + "-" * 20 + "+"
  print(linea)
  print(f"| {'Tipo':<10} | {'Monto':>14} | {'Saldo resultante':>18} |")
  print(linea)

  for movimiento in cuenta["movimientos"]:
    if movimiento['monto'] is None:
      monto_texto = "-"
    else:
      monto_texto = f"${movimiento['monto']:,}"
    saldo_texto = f"${movimiento['saldo resultante']:,}"
    print(f"| {movimiento['tipo']:<10} | {monto_texto:>14} | {saldo_texto:>18} |")

  print(linea)

print("Bienvenid@", cuenta['titular'])
operacion(cuenta, menu())
```
